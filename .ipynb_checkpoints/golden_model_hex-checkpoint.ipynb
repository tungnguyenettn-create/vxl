# Golden Model from HEX files — Zero NumPy

This notebook implements a **pure Python golden model** for a 5-layer pow2-weight CNN inference engine.

## What this notebook does

- Loads **all parameters and weights** directly from pre-generated HEX files in `golden_out/`
- Performs **all arithmetic using pure Python integers** — no NumPy, no PyTorch, no external libraries
- Walks through every step of inference so a learner can trace exactly what happens to data flowing through the hardware

## Why pure Python integers?

The hardware does **fixed-point integer arithmetic**. Python's native `int` is unbounded (no overflow), which makes it perfect for verifying hardware behaviour: intermediate accumulators can grow large before the requantisation step clips them back to range.

## Notebook flow

| Step | Description |
|------|-------------|
| 1 | Read HEX files (unsigned & signed) |
| 2 | Reshape flat lists into 2D/3D/4D tensors |
| 3 | Load weights + BN parameters |
| 4 | Load input image |
| 5 | 3×3 conv with pow2 weights (bit-shifts) |
| 6 | Requantise: BN fold + ReLU + 6-bit clamp |
| 7 | 2×2 MaxPool |
| 8 | Global Average Pool + Fully Connected layer |
| 9 | Full forward pass |
| 10 | Run on test image |
| 11 | Verify against reference HEX outputs |

````python
import os, json, math

HERE    = os.path.abspath('')
HEX_DIR = os.path.join(HERE, 'golden_out')

# Load params.json
with open(os.path.join(HEX_DIR, 'params.json')) as f:
    params = json.load(f)

IN_BITS  = params['in_bits']    # 4  — input pixels are 4-bit
ABITS    = params['abits']      # 6  — activations are 6-bit
W_FRAC   = params['w_frac']     # 7  — weight fractional bits
M_BITS   = params['m_bits']     # 16 — BN multiplier bit width
BQ_BITS  = params['bq_bits']    # 30 — BN bias bit width
AMAX     = (1 << ABITS) - 1    # 63 — maximum 6-bit activation value
INMAX    = (1 << IN_BITS) - 1  # 15 — maximum 4-bit input value
SH       = params['SH']         # list[int]: per-layer arithmetic right-shift amount
pools    = params['pools']      # list[bool]: True = apply 2x2 maxpool after layer
kept_idx = params['kept_idx']   # list[list[int]]: surviving channel indices per layer
T        = params['T']          # int: FC decision threshold
nL       = len(SH)              # number of conv layers (5)

print('Params loaded from params.json:')
print(f'  Layers : {nL}')
print(f'  Pools  : {pools}')
print(f'  IN_BITS={IN_BITS}, ABITS={ABITS}, W_FRAC={W_FRAC}')
print(f'  M_BITS={M_BITS}, BQ_BITS={BQ_BITS}, AMAX={AMAX}')
print(f'  Decision threshold T={T}')
print(f'  Kept channels per layer: {[len(k) for k in kept_idx]}')
print(f'  SH values: {SH}')
````

## Step 1: Reading HEX files

Every HEX file in `golden_out/` stores one value per line in **hexadecimal**.

### Unsigned vs signed values

Some values (pixel intensities, 4-bit weight codes, 6-bit activations) are naturally **unsigned** — we just parse them with `int(x, 16)`.

Other values (BN multipliers, BN biases, FC accumulators) are **signed integers** stored in **two's complement**.
For an `N`-bit two's complement number:
- If the value is **≥ 2^(N-1)**, it represents a **negative** number: subtract `2^N` to get the signed value.
- Otherwise it is positive and can be used as-is.

**Example** — 16-bit signed:
```
hex: FFFF  → unsigned: 65535  → 65535 ≥ 32768  → signed: 65535 - 65536 = -1
hex: 7FFF  → unsigned: 32767  → 32767 < 32768  → signed: 32767  (no change)
```

**Example** — 30-bit signed:
```
hex: 20000000 → unsigned: 536870912 → ≥ 2^29=536870912 → negative
              → signed: 536870912 - 2^30 = -536870912
```

Python's `int` type has no bit-width, so we implement sign extension manually.

````python
def read_hex_unsigned(path):
    """Read a hex file, one value per line. Returns list of int (unsigned)."""
    with open(path) as f:
        lines = [l.strip() for l in f if l.strip()]
    return [int(x, 16) for x in lines]

def sign_extend(value, bits):
    """
    Convert an unsigned integer to signed using two's complement.
    For an N-bit representation:
      - Values < 2^(N-1) are positive  (no change)
      - Values >= 2^(N-1) are negative (subtract 2^N)
    """
    if value >= (1 << (bits - 1)):
        value -= (1 << bits)
    return value

def read_hex_signed(path, bits):
    """Read a hex file as signed integers (two's complement, N-bit)."""
    values = read_hex_unsigned(path)
    return [sign_extend(v, bits) for v in values]

# Quick sanity check on the first image
sample = read_hex_unsigned(os.path.join(HEX_DIR, 'img0_in.hex'))
print(f'img0_in.hex: first 8 values = {sample[:8]}')
print(f'Total pixels: {len(sample)} (should be 64x64 = {64*64})')
print(f'Pixel value range: [{min(sample)}, {max(sample)}]  (expect [0, {INMAX}])')
````

## Step 2: Reshape flat lists into 2D / 3D / 4D arrays

HEX files store data as a **flat sequence** of values in row-major (C) order.
We need to interpret that flat list as a multi-dimensional tensor.

The layout conventions used in this model are:

| Data | Layout | Flat index formula |
|------|--------|--------------------|
| Image / feature map | `[channel][row][col]` | `c*H*W + r*W + col` |
| Conv weights | `[cout][cin][ky][kx]` | `co*(cin*KH*KW) + ci*(KH*KW) + ky*KW + kx` |
| FC weights | `[class][cin]` | `cl*cin + ci` |

We build **nested Python lists** so indexing looks like `tensor[c][r][col]`.

No NumPy arrays — just `list` of `list` of `list` ...

````python
def make_grid(flat, H, W):
    """Reshape flat list into H x W nested list (2D grid)."""
    assert len(flat) == H * W, f'Expected {H*W} values, got {len(flat)}'
    return [[flat[r * W + c] for c in range(W)] for r in range(H)]

def make_tensor3d(flat, C, H, W):
    """Reshape flat list into C x H x W nested list."""
    assert len(flat) == C * H * W, f'Expected {C*H*W} values, got {len(flat)}'
    return [[[flat[c * H * W + r * W + col]
              for col in range(W)]
             for r in range(H)]
            for c in range(C)]

def make_tensor4d(flat, C_out, C_in, KH, KW):
    """Reshape flat list into C_out x C_in x KH x KW nested list."""
    total = C_out * C_in * KH * KW
    assert len(flat) == total, f'Expected {total} values, got {len(flat)}'
    stride_co = C_in * KH * KW
    stride_ci = KH * KW
    stride_kh = KW
    return [[[[flat[co * stride_co + ci * stride_ci + kh * stride_kh + kw]
               for kw in range(KW)]
              for kh in range(KH)]
             for ci in range(C_in)]
            for co in range(C_out)]

# Verify: reshape a small example and check round-trip
flat_test = list(range(24))  # 2 x 3 x 4
t = make_tensor3d(flat_test, 2, 3, 4)
print('make_tensor3d test (2x3x4):')
print(f'  t[0][0] = {t[0][0]}  (expect [0,1,2,3])')
print(f'  t[1][2] = {t[1][2]}  (expect [20,21,22,23])')
print('Helper functions defined: make_grid, make_tensor3d, make_tensor4d')
````

## Step 3: Load weights and BN parameters

### Weight encoding — 4-bit code `[s e2 e1 e0]`

Each weight is stored as a 4-bit code:

| Bit | Name | Meaning |
|-----|------|---------|
| 3 | `s` | Sign bit: 0 = positive, 1 = negative |
| 2:0 | `e` | Exponent (3 bits) |

Decoding rule:
- If `e == 0` → weight is **zero**
- Otherwise → `|w| = 2^(-e)`, integer form: `w_int = ±2^(W_FRAC - e)` where `W_FRAC = 7`

**Examples** (`W_FRAC = 7`):

| code | s | e | w_int |
|------|---|---|-------|
| `0b0001` | 0 | 1 | `+2^6 = +64` |
| `0b0010` | 0 | 2 | `+2^5 = +32` |
| `0b0111` | 0 | 7 | `+2^0 = +1`  |
| `0b1001` | 1 | 1 | `-2^6 = -64` |
| `0b0000` | 0 | 0 | `0`           |
| `0b1000` | 1 | 0 | `0`           |

In hardware, multiplying by `w_int = ±2^(7-e)` is a **single bit-shift**, which is extremely cheap.

### BN parameters

After convolution the accumulator is in units of `2^(-7)` (because `w_int` has 7 fractional bits).
Batch normalisation is **folded** into two integer parameters per output channel:
- `M[co]` — 16-bit signed multiplier
- `Bq[co]` — 30-bit signed bias

````python
def decode_weight_code(code):
    """
    Decode a 4-bit pow2 weight code [s e2 e1 e0] to an integer weight.

    Bit layout:
      bit 3  = s (sign: 0=positive, 1=negative)
      bits 2:0 = e (exponent, 3 bits)

    Result:
      e == 0  ->  0
      e != 0  ->  (-1)^s * 2^(W_FRAC - e)   where W_FRAC=7
    """
    s = (code >> 3) & 1   # sign bit
    e = code & 7           # 3-bit exponent
    if e == 0:
        return 0
    magnitude = 1 << (W_FRAC - e)   # 2^(7-e)
    return -magnitude if s else magnitude

# Show the full decode table so the pattern is clear
print('Weight decode table (all 16 codes):')
print(f'  {"code":>4}  {"binary":>6}  {"s":>2}  {"e":>2}  {"w_int":>6}')
for code in range(16):
    s = (code >> 3) & 1
    e = code & 7
    w = decode_weight_code(code)
    print(f'  {code:>4}  {bin(code)[2:].zfill(4):>6}  {s:>2}  {e:>2}  {w:>6}')

# ---------------------------------------------------------------------------
# Load conv weights for each layer
# ---------------------------------------------------------------------------
W_int  = []   # decoded integer weights: list of 4D arrays [cout][cin][3][3]
W_code = []   # raw 4-bit codes:         list of 4D arrays [cout][cin][3][3]

print()
for i in range(nL):
    cout = len(kept_idx[i])
    cin  = 1 if i == 0 else len(kept_idx[i - 1])

    codes_flat = read_hex_unsigned(os.path.join(HEX_DIR, f'w_L{i}.hex'))
    codes_4d   = make_tensor4d(codes_flat, cout, cin, 3, 3)
    W_code.append(codes_4d)

    # Decode every code to its integer weight
    wint_4d = [[[[decode_weight_code(codes_4d[co][ci][ky][kx])
                  for kx in range(3)]
                 for ky in range(3)]
                for ci in range(cin)]
               for co in range(cout)]
    W_int.append(wint_4d)

    # Sample: show first kernel of first output channel
    sample_kernel = codes_4d[0][0]  # shape [3][3] of codes
    sample_wint   = wint_4d[0][0]   # shape [3][3] of integers
    print(f'Layer {i}: weight shape [{cout}][{cin}][3][3]')
    print(f'  code[0][0] kernel : {sample_kernel}')
    print(f'  wint[0][0] kernel : {sample_wint}')

# ---------------------------------------------------------------------------
# Load FC weights
# ---------------------------------------------------------------------------
n_classes = 2
n_fc_in   = len(kept_idx[-1])

fc_codes_flat = read_hex_unsigned(os.path.join(HEX_DIR, 'fc_w.hex'))
# FC layout: [class][cin], stored row-major
FC_code = [[fc_codes_flat[cl * n_fc_in + ci] for ci in range(n_fc_in)]
            for cl in range(n_classes)]
FC_int  = [[decode_weight_code(FC_code[cl][ci]) for ci in range(n_fc_in)]
            for cl in range(n_classes)]
print(f'\nFC: weight shape [{n_classes}][{n_fc_in}]')
print(f'  FC_int[0][:5] = {FC_int[0][:5]}')
print(f'  FC_int[1][:5] = {FC_int[1][:5]}')

# ---------------------------------------------------------------------------
# Load BN M and Bq for each layer
# ---------------------------------------------------------------------------
BN_M  = []   # per-layer list[cout], signed 16-bit
BN_Bq = []   # per-layer list[cout], signed BQ_BITS-bit

print()
for i in range(nL):
    cout = len(kept_idx[i])
    m_vals  = read_hex_signed(os.path.join(HEX_DIR, f'bn_M_L{i}.hex'),  M_BITS)
    bq_vals = read_hex_signed(os.path.join(HEX_DIR, f'bn_B_L{i}.hex'),  BQ_BITS)
    assert len(m_vals)  == cout, f'Layer {i}: expected {cout} M values, got {len(m_vals)}'
    assert len(bq_vals) == cout, f'Layer {i}: expected {cout} Bq values, got {len(bq_vals)}'
    BN_M.append(m_vals)
    BN_Bq.append(bq_vals)
    print(f'Layer {i}: BN M[0]={m_vals[0]:8d}, Bq[0]={bq_vals[0]:12d}  (cout={cout})')

print('\nAll weights and BN parameters loaded!')
````

## Step 4: Load input image

The CNN operates on **4-bit pixel values** (range 0–15).

In a real pipeline the 8-bit pixel `p` is converted to 4 bits by taking the top nibble: `p >> 4`.
The HEX file stores these 4-bit values directly, so we just read them as unsigned integers.

Layout: row-major, `[row][col]`, 64×64 = 4096 values total.

We select the first test image from `params.json['images']`, which also tells us the ground-truth label.

````python
# Select the first test image described in params.json
IMG_IDX = params['images'][0]['idx']    # e.g. 0
LABEL   = params['images'][0]['label']  # 0 = no_person, 1 = person

in_flat = read_hex_unsigned(os.path.join(HEX_DIR, f'img{IMG_IDX}_in.hex'))
H_in = W_in = 64
img_in = make_grid(in_flat, H_in, W_in)   # nested list [H][W], values 0..15

label_str = 'person' if LABEL else 'no_person'
print(f'Image index : {IMG_IDX}')
print(f'Label       : {LABEL} ({label_str})')
print(f'Shape       : {H_in} x {W_in}')
print(f'Value range : [{min(in_flat)}, {max(in_flat)}]  (expect [0, {INMAX}])')
print()
print('Top-left 8x8 corner of the image:')
for row in img_in[:8]:
    print(' ', ' '.join(f'{v:2d}' for v in row[:8]))
````

## Step 5: 3×3 Convolution with pow2 weights (bit-shifts)

### Formula

For each output channel `co` and spatial position `(y, x)`:

$$
\text{acc}[co][y][x] = \sum_{ci} \sum_{ky=0}^{2} \sum_{kx=0}^{2}
    \text{in}[ci][y+ky][x+kx] \times w_{\text{int}}[co][ci][ky][kx]
$$

### Padding

We use **same padding** (pad = 1 on each side), so the output spatial dimensions equal the input.  
Zero-pad the input to `(H+2) × (W+2)` before the loops.

### Why accumulators carry fractional bits

Because `w_int = ±2^(7-e)`, every product `in * w_int` has an implicit factor of `2^(-7)` relative to the true real-valued weight.  
The accumulator therefore represents the true result scaled by `2^7`.  
This scaling is absorbed into the BN multiplier `M` in the requantisation step.

### Hardware view

- Multiply by `w_int = ±2^(7-e)` → **bit-shift left by `(7-e)`, negate if `s=1`**
- Zero weights (`e=0`) → skip the MAC entirely
- The accumulator for one output pixel sums up to `cin × 9` such shifted values

````python
def arith_shift_right(x, n):
    """
    Arithmetic right shift by n bits.
    Python's >> operator already performs arithmetic shift for integers
    (floor division), so negative numbers shift correctly without extra work.

    Examples:
      arith_shift_right( 10, 2) =  2  (10 // 4)
      arith_shift_right(-10, 2) = -3  (floor(-10/4) = floor(-2.5) = -3)
    """
    return x >> n

def conv3x3_int(inp, w_int, cin, cout, H, W):
    """
    3x3 convolution with same padding (pad=1 on each side).

    Args:
        inp   : list[cin][H][W]        — integer input activations
        w_int : list[cout][cin][3][3]  — integer weights (w_int = ±2^(7-e))
        cin, cout, H, W               — dimensions

    Returns:
        out   : list[cout][H][W]       — raw accumulator (integer, 7 implicit frac bits)

    The output is NOT requantised yet — values can be large and negative.
    Requantisation (BN + ReLU + 6-bit clamp) happens in the next step.
    """
    # Zero-pad input: add 1 row/col of zeros on every side
    H_pad = H + 2
    W_pad = W + 2
    inp_pad = [[[0] * W_pad for _ in range(H_pad)] for _ in range(cin)]
    for c in range(cin):
        for r in range(H):
            for col in range(W):
                inp_pad[c][r + 1][col + 1] = inp[c][r][col]

    # 3x3 convolution
    out = [[[0] * W for _ in range(H)] for _ in range(cout)]
    for co in range(cout):
        for r in range(H):
            for col in range(W):
                acc = 0
                for ci in range(cin):
                    for ky in range(3):
                        for kx in range(3):
                            pixel = inp_pad[ci][r + ky][col + kx]
                            w     = w_int[co][ci][ky][kx]
                            # In hardware: w = ±2^(7-e), so this is ±(pixel << (7-e))
                            acc  += pixel * w
                out[co][r][col] = acc
    return out

# Quick demonstration: convolve a 2x2 patch manually vs the function
print('arith_shift_right(-10, 2) =', arith_shift_right(-10, 2), ' (expect -3: floor(-2.5))')
print('arith_shift_right( 10, 2) =', arith_shift_right( 10, 2), ' (expect  2)')
print('conv3x3_int defined')
````

## Step 6: Requantise — BN fold + ReLU + 6-bit clamp

After convolution the accumulator is a large integer.  
We need to bring it back to 6-bit range `[0, 63]` for the next layer.

### Formula (per output channel `co`, per spatial position)

$$
q = \text{clamp}\!\left(\left\lfloor \frac{\text{acc} \times M_{co} + Bq_{co}}{2^{SH}} \right\rfloor,\ 0,\ 63\right)
$$

Written in code:
```python
val = arith_shift_right(acc * M + Bq, SH)
q   = max(0, min(63, val))
```

### What each term does

| Term | Role |
|------|------|
| `acc * M` | Applies BN gain + undoes the `2^7` weight scaling |
| `+ Bq` | Applies BN bias (folded, pre-rounded) |
| `>> SH` | Arithmetic right-shift to bring back to activation range |
| `clamp(·, 0, 63)` | ReLU (clips negative values to 0) + 6-bit saturation |

### Why arithmetic right-shift and not integer division?

For negative numbers, Python's `//` and `>>` both implement **floor division** (round toward −∞).  
This matches the hardware's arithmetic shift register behaviour, so `x >> n` is exactly right.

````python
def requant(acc_map, M_list, Bq_list, SH_val, cout, H, W):
    """
    Requantise an accumulator map to 6-bit activations.

    Args:
        acc_map : list[cout][H][W]  — raw conv accumulators
        M_list  : list[cout]        — per-channel BN multiplier (signed 16-bit)
        Bq_list : list[cout]        — per-channel BN bias (signed 30-bit)
        SH_val  : int               — arithmetic right-shift amount for this layer
        cout, H, W                  — dimensions

    Returns:
        out : list[cout][H][W]      — 6-bit activations, values in [0, 63]
    """
    out = [[[0] * W for _ in range(H)] for _ in range(cout)]
    for co in range(cout):
        m  = M_list[co]
        bq = Bq_list[co]
        for r in range(H):
            for col in range(W):
                # 1. BN: scale accumulator and add bias
                val = acc_map[co][r][col] * m + bq
                # 2. Arithmetic right-shift to bring into 6-bit range
                val = arith_shift_right(val, SH_val)
                # 3. ReLU + 6-bit saturation: clamp to [0, 63]
                val = max(0, min(AMAX, val))
                out[co][r][col] = val
    return out

print('requant defined')

# Manual example to verify the formula
print()
print('Requant example:')
acc_ex = 12345
M_ex   = BN_M[0][0]
Bq_ex  = BN_Bq[0][0]
SH_ex  = SH[0]
val_ex = arith_shift_right(acc_ex * M_ex + Bq_ex, SH_ex)
q_ex   = max(0, min(AMAX, val_ex))
print(f'  acc={acc_ex}, M={M_ex}, Bq={Bq_ex}, SH={SH_ex}')
print(f'  (acc*M + Bq) = {acc_ex * M_ex + Bq_ex}')
print(f'  >> SH        = {val_ex}')
print(f'  clamped q    = {q_ex}')
````

## Step 7: 2×2 MaxPool

After some conv layers, spatial dimensions are halved using **2×2 non-overlapping max pooling**.

For each channel `c` and each 2×2 block at position `(r, col)` (in output coordinates):
$$
\text{out}[c][r][col] = \max\bigl(
    \text{in}[c][2r][2col],\ 
    \text{in}[c][2r][2col+1],\ 
    \text{in}[c][2r+1][2col],\ 
    \text{in}[c][2r+1][2col+1]
\bigr)
$$

Output size: `(H//2) × (W//2)`

In this model, pooling is applied after layers 0, 1, and 3 (see `pools = [True, True, False, True, False]`).

````python
def maxpool2(inp, cout, H, W):
    """
    2x2 non-overlapping max pool.

    Args:
        inp  : list[cout][H][W]           — input activations
        cout, H, W                        — input dimensions

    Returns:
        out  : list[cout][H//2][W//2]     — pooled activations
    """
    H_out = H // 2
    W_out = W // 2
    out   = [[[0] * W_out for _ in range(H_out)] for _ in range(cout)]
    for c in range(cout):
        for r in range(H_out):
            for col in range(W_out):
                r0 = r * 2
                c0 = col * 2
                val = max(
                    inp[c][r0    ][c0    ],
                    inp[c][r0    ][c0 + 1],
                    inp[c][r0 + 1][c0    ],
                    inp[c][r0 + 1][c0 + 1]
                )
                out[c][r][col] = val
    return out

print('maxpool2 defined')

# Quick demonstration: pool a 2x4 feature map (1 channel)
demo_inp = [[[1, 3, 2, 8],
             [5, 2, 7, 4]]]
demo_out = maxpool2(demo_inp, cout=1, H=2, W=4)
print('Demo pool input  [0]:', demo_inp[0])
print('Demo pool output [0]:', demo_out[0])  # expect [[5, 8]]
````

## Step 8: Global Average Pool + Fully Connected layer

### Global Average Pool (GAP)

After the last conv layer the feature map has shape `[cout_last][H][W]`.  
GAP reduces each channel to a single scalar:
$$
S_c = \sum_{r=0}^{H-1} \sum_{col=0}^{W-1} \text{act}[c][r][col]
$$

We store the **sum** (not the average) because:
- Dividing by `H*W` would require floating point or extra fixed-point scaling.
- The FC weights can absorb the `1/(H*W)` factor at training time.
- We just need `D = acc_fc[1] - acc_fc[0]` to be compared against threshold `T`.

### Fully Connected layer

With 2 output classes:
$$
\text{acc\_fc}[cl] = \sum_{c=0}^{cout_{last}-1} S_c \times w_{FC}[cl][c]
$$

FC weights are pow2-encoded exactly like conv weights.

### Decision

$$
D = \text{acc\_fc}[1] - \text{acc\_fc}[0]
$$
$$
\text{pred} = \begin{cases} 1 & \text{(person)} & \text{if } D > T \\ 0 & \text{(no person)} & \text{otherwise} \end{cases}
$$

````python
def global_avg_pool_sum(inp, cout, H, W):
    """
    Global average pool — returns the SPATIAL SUM per channel (not the mean).

    S[c] = sum_{r,col} inp[c][r][col]

    Args:
        inp        : list[cout][H][W]
        cout, H, W : dimensions

    Returns:
        S : list[cout]  — integer sums, one per channel
    """
    S = [0] * cout
    for c in range(cout):
        total = 0
        for r in range(H):
            for col in range(W):
                total += inp[c][r][col]
        S[c] = total
    return S

def fc_layer(S, fc_int, n_classes, n_in):
    """
    Fully connected layer operating on GAP spatial sums.

    acc_fc[cl] = sum_c( S[c] * fc_int[cl][c] )

    Args:
        S          : list[n_in]              — GAP sums from last conv layer
        fc_int     : list[n_classes][n_in]   — decoded integer FC weights
        n_classes  : int                     — number of output classes (2)
        n_in       : int                     — number of input channels

    Returns:
        acc_fc : list[n_classes]  — raw class scores
    """
    acc_fc = [0] * n_classes
    for cl in range(n_classes):
        acc = 0
        for c in range(n_in):
            acc += S[c] * fc_int[cl][c]
        acc_fc[cl] = acc
    return acc_fc

print('global_avg_pool_sum defined')
print('fc_layer defined')
````

## Step 9: Full forward pass

Now we wire everything together into one function that takes an input image and returns the prediction.

Data flow through the network:

```
img_in [1×64×64]
  │
  ├─ Layer 0: conv3x3 → requant → maxpool →  [8×32×32]
  ├─ Layer 1: conv3x3 → requant → maxpool →  [16×16×16]
  ├─ Layer 2: conv3x3 → requant            →  [23×16×16]
  ├─ Layer 3: conv3x3 → requant → maxpool →  [31×8×8]
  └─ Layer 4: conv3x3 → requant            →  [29×8×8]
        │
       GAP  →  S[29]  (spatial sums)
        │
        FC   →  acc_fc[2]  (class scores)
        │
   D = acc_fc[1] - acc_fc[0]  →  compare with T  →  prediction
```

*(Exact channel counts depend on `kept_idx`.)*

````python
def golden_forward(img_in, H=64, W=64):
    """
    Full CNN forward pass on a single image using pure Python integers.

    Args:
        img_in : list[H][W]  — 4-bit input pixels, values 0..15
        H, W   : int         — spatial dimensions (default 64x64)

    Returns:
        dict with keys:
          'layer_accs'    : list[nL] of list[cout][H][W]  — raw conv accumulators
          'layer_outputs' : list[nL] of list[cout][H][W]  — 6-bit activations (after requant & pool)
          'S'             : list[cout_last]               — GAP spatial sums
          'n_pixels'      : int                           — H*W at last layer
          'acc_fc'        : list[2]                       — FC class scores
          'D'             : int                           — decision value
          'pred'          : int                           — 1=person, 0=no_person
    """
    # Layer 0 input: wrap single-channel image as [1][H][W]
    act = [[img_in[r][:] for r in range(H)]]   # list[1][H][W]
    H_cur, W_cur = H, W

    layer_outputs = []
    layer_accs    = []

    for i in range(nL):
        cout = len(kept_idx[i])
        cin  = len(act)

        print(f'  Layer {i}: [{cin}x{H_cur}x{W_cur}]', end='')

        # ---- 1. 3x3 Convolution ----
        acc = conv3x3_int(act, W_int[i], cin, cout, H_cur, W_cur)
        layer_accs.append(acc)

        # ---- 2. Requantise: BN + ReLU + 6-bit clamp ----
        act = requant(acc, BN_M[i], BN_Bq[i], SH[i], cout, H_cur, W_cur)

        # ---- 3. Optional 2x2 MaxPool ----
        if pools[i]:
            act = maxpool2(act, cout, H_cur, W_cur)
            H_cur //= 2
            W_cur //= 2

        layer_outputs.append(act)
        pool_str = 'pool' if pools[i] else 'no pool'
        print(f' -> [{cout}x{H_cur}x{W_cur}]  ({pool_str})')

    # ---- 4. Global Average Pool (sum) ----
    cout_last = len(kept_idx[-1])
    S         = global_avg_pool_sum(act, cout_last, H_cur, W_cur)
    n_pixels  = H_cur * W_cur
    print(f'  GAP: [{cout_last}x{H_cur}x{W_cur}] -> S[{cout_last}]  (n_pixels={n_pixels})')

    # ---- 5. Fully Connected ----
    acc_fc = fc_layer(S, FC_int, n_classes=2, n_in=cout_last)
    print(f'  FC : acc_fc = {acc_fc}')

    # ---- 6. Decision ----
    D    = acc_fc[1] - acc_fc[0]
    pred = 1 if D > T else 0
    print(f'  Decision: D={D}, T={T}, pred={pred}')

    return {
        'layer_accs'    : layer_accs,
        'layer_outputs' : layer_outputs,
        'S'             : S,
        'n_pixels'      : n_pixels,
        'acc_fc'        : acc_fc,
        'D'             : D,
        'pred'          : pred,
    }

print('golden_forward defined')
````

## Step 10: Run the forward pass on the test image

We now call `golden_forward` with the image loaded in Step 4.  
Watch the per-layer print statements to see dimensions shrink as pooling is applied.

````python
print(f'Running golden model on image {IMG_IDX}  (label={LABEL})...')
print()
result = golden_forward(img_in)
print()

label_str = 'person (pred=1)' if result['pred'] == 1 else 'no_person (pred=0)'
ref_D     = params['images'][0].get('D', 'N/A')

print('=' * 50)
print(f"Result for image {IMG_IDX} (true label={LABEL}):")
print(f"  acc_fc          = {result['acc_fc']}")
print(f"  D = acc_fc[1] - acc_fc[0] = {result['D']}")
print(f"  T (threshold)   = {T}")
print(f"  Prediction      : {label_str}")
if ref_D != 'N/A':
    match = result['D'] == ref_D
    print(f"  Expected D (params.json): {ref_D}")
    print(f"  D match                 : {match}")
print('=' * 50)
````

## Step 11: Verify against reference HEX files

The `golden_out/` directory contains reference outputs generated by the hardware simulation:

- `img{k}_out_L{i}.hex` — 6-bit activation output for layer `i` (after requant + optional pool)
- `img{k}_fc.hex` — GAP sums `S`, FC accumulators `acc_fc[0]`, `acc_fc[1]`, then decision `D`

We compare our Python results against these references element-by-element.

If all values match, our pure-Python model is **bit-exact** with the hardware model. ✓

````python
def verify_layer_output(pred_out, ref_flat, cout, H, W, layer_idx):
    """
    Compare a predicted output tensor against a flat reference list.

    Flattens pred_out in [cout][H][W] order (matching the hex file layout)
    and reports any mismatches.

    Returns True if all values match.
    """
    # Flatten predicted output in the same order as the hex file
    pred_flat = []
    for c in range(cout):
        for r in range(H):
            for col in range(W):
                pred_flat.append(pred_out[c][r][col])

    if len(pred_flat) != len(ref_flat):
        print(f'  Layer {layer_idx}: SIZE MISMATCH — got {len(pred_flat)}, expected {len(ref_flat)}')
        return False

    mismatches = 0
    for idx, (p, r) in enumerate(zip(pred_flat, ref_flat)):
        if p != r:
            mismatches += 1
            if mismatches <= 5:  # show at most 5 mismatches to keep output tidy
                # Decode index to [c][row][col]
                col_ = idx % W
                row_ = (idx // W) % H
                ch_  = idx // (H * W)
                print(f'    Mismatch at [c={ch_}, r={row_}, col={col_}] '
                      f'(flat idx {idx}): got {p}, expected {r}')

    total = len(ref_flat)
    print(f'  Layer {layer_idx}: {total - mismatches}/{total} match '
          f'({100 * (total - mismatches) / total:.1f}%)')
    return mismatches == 0

# ---------------------------------------------------------------------------
# Verify layer outputs
# ---------------------------------------------------------------------------
print(f'Verifying layer outputs for image {IMG_IDX}...')
H_ref, W_ref = 64, 64
all_ok = True

for i in range(nL):
    cout = len(kept_idx[i])
    if pools[i]:
        H_ref //= 2
        W_ref //= 2

    ref_path = os.path.join(HEX_DIR, f'img{IMG_IDX}_out_L{i}.hex')
    if not os.path.exists(ref_path):
        print(f'  Layer {i}: reference file not found, skipping')
        continue

    ref_flat = read_hex_unsigned(ref_path)
    ok = verify_layer_output(result['layer_outputs'][i], ref_flat, cout, H_ref, W_ref, i)
    all_ok = all_ok and ok

# ---------------------------------------------------------------------------
# Verify FC output
# ---------------------------------------------------------------------------
print()
fc_ref_path = os.path.join(HEX_DIR, f'img{IMG_IDX}_fc.hex')
if os.path.exists(fc_ref_path):
    ref_fc_flat = read_hex_unsigned(fc_ref_path)

    # Layout: S[0..cout_last-1], acc_fc[0], acc_fc[1], D
    # (derived from: np.r_[t['S'][0], t['acc_fc'][0], t['D']] in the original notebook)
    # All values stored as 32-bit two's complement
    cout_last = len(kept_idx[-1])

    ref_S   = [sign_extend(v, 32) for v in ref_fc_flat[:cout_last]]
    ref_acc = [sign_extend(v, 32) for v in ref_fc_flat[cout_last : cout_last + 2]]
    ref_D   = sign_extend(ref_fc_flat[cout_last + 2], 32)

    # Compare S values
    s_mismatches = sum(1 for a, b in zip(result['S'], ref_S) if a != b)
    print(f'GAP S verification: {cout_last - s_mismatches}/{cout_last} match')
    if s_mismatches > 0:
        for idx, (a, b) in enumerate(zip(result['S'], ref_S)):
            if a != b:
                print(f'  S[{idx}]: got {a}, expected {b}')
                break

    print(f'FC  acc_fc : our={result["acc_fc"]}, ref={ref_acc}')
    print(f'FC  D      : our={result["D"]}, ref={ref_D}, match={result["D"] == ref_D}')
    all_ok = all_ok and (result['D'] == ref_D)
else:
    print(f'FC reference file not found: {fc_ref_path}')

print()
if all_ok:
    print('✓ ALL outputs match the reference HEX files — model is bit-exact!')
else:
    print('✗ Some outputs do not match. Check the mismatches above.')
````

## Summary

### What we computed

| Step | Operation | Key detail |
|------|-----------|------------|
| Load | Read HEX → Python lists | Sign-extend two's complement values |
| Weights | 4-bit code → integer | `w_int = ±2^(7-e)` |
| Conv | 3×3 same-padding conv | Accumulator has 7 implicit fractional bits |
| Requant | `clamp((acc*M + Bq) >> SH, 0, 63)` | BN + ReLU + 6-bit clamp |
| MaxPool | 2×2 non-overlapping max | Applied after layers 0, 1, 3 |
| GAP | Sum over H×W per channel | Integer sum, no division |
| FC | dot product with pow2 weights | Same encoding as conv |
| Decision | `D = acc_fc[1] - acc_fc[0] > T` | Signed comparison |

### Hardware mapping

Every Python operation here maps directly to cheap hardware primitives:

| Python | Hardware |
|--------|----------|
| `pixel * w_int` where `w_int = ±2^k` | Barrel shifter (shift by `k`, negate if negative sign) |
| `acc += ...` | Adder tree / accumulator register |
| `acc * M + Bq` | Multiply-accumulate (MAC) with 16-bit multiplier |
| `>> SH` | Arithmetic shift register |
| `max(0, min(63, val))` | Two comparators (clamp low / clamp high) |
| `max(a, b, c, d)` | 4-input comparator tree |
| `sum(...)` over GAP | Accumulator |
| `acc_fc[1] - acc_fc[0] > T` | Subtractor + comparator |

### Key insight

By restricting weights to powers of two, the **most expensive operation in a CNN — multiply — becomes a bit-shift**.  
This notebook shows, in transparent Python, exactly what the hardware computes at every step.
